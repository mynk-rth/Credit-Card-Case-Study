### a. In case age is less than 18, replace it with mean of age values. 

```python

mean_age = int(Customer['Age'].mean())
Customer.loc[Customer['Age'] < 18, 'Age'] = mean_age
Customer
```

### b. In case spend amount is more than the limit, replace it with 50% of that customer’s limit.  (customer’s limit provided in acquisition table is the per transaction limit on his card) 

Merging Customer Acqusition and spend 

```python
Cust_spend=pd.merge(Customer,Spend)
Cust_spend
```

Replace spend amount with 50% of customer limit in case of crossing customer limit

```python
Cust_spend.loc[Cust_spend['Amount']>Cust_spend['Limit'],'Amount']= 0.5*Cust_spend['Limit']
Cust_spend
```

## c. Incase the repayment amount is more than the limit, replace the repayment with the limit.

```python
Cust_repay=pd.merge(Customer,Repayment)
Cust_repay
```

Replace repayment amount with customer limit in case of crossing customer limit

```python
Cust_repay.loc[Cust_repay['Amount']>Cust_repay['Limit'],'Amount']= Cust_repay['Limit']
Cust_repay
```

### c. What is the average monthly spend by customers? 

Find month name for grouping 

```python
Cust_spend['Month_name']=pd.to_datetime(Cust_spend['Month'],dayfirst=True,format='mixed').dt.month_name()
Cust_spend
```

```python
Avg_spend=Cust_spend['Amount'].groupby(Cust_spend['Month_name']).mean()

month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]
Avg_spend = Avg_spend.reindex(month_order)
Avg_spend
```

### d. What is the average monthly repayment by customers? 

```python
Cust_repay['Month_name']=pd.to_datetime(Cust_repay['Month'],dayfirst=True,format='mixed').dt.month_name()
Cust_repay
```

```python
Avg_repay=Cust_repay['Amount'].groupby(Cust_repay['Month_name']).mean()
Avg_repay = Avg_repay.reindex(month_order)
Avg_repay
```

### e. If the monthly rate of interest is 2.9%, what is the profit for the bank for each month? 
(Profit is defined as interest earned on Monthly Profit. Monthly Profit = Monthly repayment  – Monthly spend. Interest is earned only on positive profits and not on negative amounts) 

```python
Monthly_spend = Cust_spend.groupby('Month_name')['Amount'].sum()
Monthly_spend
```

```python
Monthly_repay= Cust_repay.groupby('Month_name')['Amount'].sum()
Monthly_repay
```

Merge month wise total of spend and repayment

```python
Spend_repay = pd.merge(Monthly_spend,Monthly_repay,left_on='Month_name',right_on='Month_name')
Spend_repay.columns = ['Spend','Repayment']
Spend_repay=Spend_repay.reindex(month_order)
Spend_repay=Spend_repay.reset_index()
Spend_repay
```

Add new column profit and interest

```python
Spend_repay['Profit'] = Spend_repay['Repayment'] - Spend_repay['Spend']
Spend_repay['Interest'] = Spend_repay.loc[Spend_repay['Profit']>0,'Profit'] * 2.9/100
Spend_repay
```

## 3. Calculate the city wise spend on each product on yearly basis. Also include a graphical representation for the same. 

Method 1 :

```python
Cust_spend['Year']=pd.to_datetime(Cust_spend['Month'],dayfirst=True,format='mixed').dt.year
Cust_spend
```

Group city, product, year

```python
city_prod=Cust_spend.groupby(['City','Product','Year'])['Amount'].sum().reset_index()
city_prod
```

Bar Graph 

```python
city_prod= city_prod.pivot(
    index='Year',
    columns=['City', 'Product'],
    values='Amount'
)

city_prod.plot(kind='bar', figsize=(14,7))

plt.title('City-wise Product Spend on Yearly Basis')
plt.xlabel('Year')
plt.ylabel('Total Spend')
plt.show()
```

Method 2 :

```python
city_prod=pd.crosstab(
    [Cust_spend['City'], Cust_spend['Product']],
    Cust_spend['Year'],
    values=Cust_spend['Amount'],
    aggfunc='sum'
)
city_prod
```

```python
city_prod.plot(
    kind='bar',
    figsize=(14,7)
)

plt.title('City-wise Product Spend on Yearly Basis')
plt.xlabel('City / Product')
plt.ylabel('Total Spend')
plt.xticks(rotation=45)
plt.show()
```

```python
city_prod.T.plot(
    kind='bar',
    figsize=(14, 7)
)

plt.title('Year-wise City and Product Spend')
plt.xlabel('Year')
plt.ylabel('Total Spend')

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

## Q5.
Write user defined PYTHON function to perform the following analysis:
 
You need to find top 10 customers for each city in terms of their repayment amount by 
different products and by different time periods i.e. year or month. The user should be able 
to specify the product (Gold/Silver/Platinum) and time period (yearly or monthly) and the 
function should automatically take these inputs while identifying the top 10 customers. 

```python
Cust_repay['Year']=pd.to_datetime(Cust_repay['Month'],dayfirst=True,format='mixed').dt.year
Cust_repay
```

```python
product = "Gold"
period_type = "Monthly"
period = "January"

if period_type == "Yearly":
    filter = Cust_repay.query(
        f'Product == "{product}" and Year == {period}'
    ).groupby(['City', 'Customer'])['Amount'].sum()

elif period_type == "Monthly":
    filter = Cust_repay.query(
        f'Product == "{product}" and Month_name == "{period}"'
    ).groupby(['City', 'Customer'])['Amount'].sum()

filter
```

```python
top10 = filter.groupby(level='City').nlargest(10)
print(top10)
```
